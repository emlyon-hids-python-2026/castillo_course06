## Objective 1


```python
# Imports
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from openpyxl.drawing.image import Image
```

```python
# Paths
# Works from the project root (uv run) and from scripts/ or notebooks/ (Interactive Window)
ROOT = Path.cwd() if (Path.cwd() / "data").is_dir() else Path.cwd().parent
DATA_DIR = ROOT / "data"
OUTPUT_DIR = DATA_DIR / "output_v1"
OUTPUT_DIR.mkdir(exist_ok=True)
```

```python
# Loading datasets
sales_part1 = pd.read_csv(DATA_DIR / "../data/Sales_MedicStore_part1.csv", sep=";")
sales_part2 = pd.read_csv(DATA_DIR / "../data/Sales_MedicStore_part2.csv", sep=";")
mapping = pd.read_excel(DATA_DIR / "../data/Mapping.xlsx")
mapping
```

```python
# Assembling sales 1 and 2
sales = pd.concat([sales_part1, sales_part2], ignore_index=True)
```

```python
# Assembling the sales table with the mapping table
sales = sales.merge(mapping, on="Sub-Category", how="left")
sales.head()
```

```python
# Calculating the total sales
sales["Total Sales"] = sales["Quantity"] * sales["Unit Price"]
sales.head()
```

```python
# Exporting the dataset to Excel
sales.to_excel(OUTPUT_DIR / "total_sales.xlsx", index=False)
```

### Questions from the operational team


```python
# Number of orders in the final dataset
nb_orders = len(sales)
nb_orders
```

```python
# Amount of total sales
df_total_sales = sales["Total Sales"].sum()
print(f"The total sales amount is: {df_total_sales}")
```

```python
# Top 3 customers by order count
customers_order_count = (
    sales.groupby("Customer Name")["Order ID"]
    .count()
    .sort_values(ascending=False)
    .head(3)
)
print("Top 3 customers by order count: ")
print(customers_order_count)
```

```python
# Total sales for France
total_sales_france = sales[sales["Country"] == "France"]["Total Sales"].sum()
print(f"The total sales amount for France is: {total_sales_france}")
```

```python
# Ratio of total sales for France compared to the total global sales
ratio_sales_france = total_sales_france / df_total_sales
print(
    f"The total sales amount for France is {ratio_sales_france:.2%} of the total global sales."
)
```

```python
# Store the answers in a dictionary
# %% Create a dictionary to store the analysis results


dict_analysis = {
    "how many lines are in Sales_MedicStore_part1.csv": nb_lines_sales_part1,
    "how many lines are in Sales_MedicStore_part2.csv": nb_lines_sales_part2,
    "how many lines are in Mapping.xlsx": nb_lines_mapping,
    "how many orders do we have": nb_orders,
    "what is the total sales amount": df_total_sales,
    "What is the total Sales for France": total_sales_france,
    "How much does the total Sales for France represent compare to the total global Sales": ratio_sales_france,
}


# Convert the dictionary to a DataFrame for better visualization
df_analysis = pd.DataFrame(list(dict_analysis.items()), columns=["Question", "Answer"])
df_analysis
```

```python
# Export tables
with pd.ExcelWriter(OUTPUT_DIR / "sales_analysis.xlsx") as writer:
    df_shapes.to_excel(writer, sheet_name="technical_q1_shapes", index=False)
    df_describe.to_excel(writer, sheet_name="technical_q5_describe")
    df_analysis.to_excel(writer, sheet_name="business_questions", index=False)


print(f"Analysis exported to {OUTPUT_DIR}")
```
